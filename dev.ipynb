```python
import json, re, os
import chromadb
from sentence_transformers import SentenceTransformer
from findoc_qa import parse_pdf, chunk_pages, search, rag_answer

pages = parse_pdf("data/pdfs/apple_10k_2024.pdf")
embed_model = SentenceTransformer("BAAI/bge-small-en-v1.5")
client = chromadb.PersistentClient(path="data/chroma_db")
collection = client.get_collection("findoc")
print(f"✅ Loaded: {len(pages)} pages, collection: {collection.count()} chunks")
```

```python
# Cell 2: Hybrid Search — BM25 + Dense 融合
from collections import Counter
import math

# ---- BM25 实现 ----
class BM25:
    def __init__(self, documents, k1=1.5, b=0.75):
        self.k1 = k1
        self.b = b
        self.docs = documents
        self.doc_len = [len(d.split()) for d in documents]
        self.avgdl = sum(self.doc_len) / len(self.doc_len)
        self.N = len(documents)
        # 建立倒排索引
        self.df = Counter()
        self.doc_freqs = []
        for doc in documents:
            tokens = set(doc.lower().split())
            self.df.update(tokens)
            self.doc_freqs.append(Counter(doc.lower().split()))

    def score(self, query):
        query_tokens = query.lower().split()
        scores = []
        for i in range(self.N):
            s = 0
            for qt in query_tokens:
                if qt not in self.df:
                    continue
                idf = math.log((self.N - self.df[qt] + 0.5) / (self.df[qt] + 0.5) + 1)
                tf = self.doc_freqs[i].get(qt, 0)
                denom = tf + self.k1 * (1 - self.b + self.b * self.doc_len[i] / self.avgdl)
                s += idf * (tf * (self.k1 + 1)) / denom
            scores.append(s)
        return scores

# 加载 chunks
with open("data/chunks.json", encoding="utf-8") as f:
    chunks = json.load(f)

texts = [c["text"] for c in chunks]
print("Building BM25 index...")
bm25 = BM25(texts)
print(f"✅ BM25 index built: {len(texts)} documents")

# ---- Hybrid Search 函数 ----
def hybrid_search(query, collection, embed_model, bm25, chunks,
                  top_k=5, dense_weight=0.5):
    """融合 dense (embedding) 和 sparse (BM25) 检索"""
    # Dense 检索
    query_emb = embed_model.encode([query]).tolist()
    dense_results = collection.query(query_embeddings=query_emb, n_results=min(top_k * 3, len(chunks)))

    # 构建 dense 分数 (归一化到 0-1，1=最相关)
    dense_scores = {}
    for doc_id, dist in zip(dense_results["ids"][0], dense_results["distances"][0]):
        dense_scores[doc_id] = 1 - dist  # cosine similarity

    # BM25 检索
    bm25_raw = bm25.score(query)
    max_bm25 = max(bm25_raw) if max(bm25_raw) > 0 else 1
    bm25_scores = {chunks[i]["id"]: bm25_raw[i] / max_bm25 for i in range(len(chunks))}

    # 融合
    all_ids = set(dense_scores.keys()) | set(k for k, v in bm25_scores.items() if v > 0)
    fused = {}
    for doc_id in all_ids:
        d = dense_scores.get(doc_id, 0)
        b = bm25_scores.get(doc_id, 0)
        fused[doc_id] = dense_weight * d + (1 - dense_weight) * b

    # 排序取 top_k
    ranked = sorted(fused.items(), key=lambda x: x[1], reverse=True)[:top_k]

    # 构建返回结果
    chunk_map = {c["id"]: c for c in chunks}
    results = {
        "ids": [[r[0] for r in ranked]],
        "documents": [[chunk_map[r[0]]["text"] for r in ranked]],
        "distances": [[1 - r[1] for r in ranked]],  # 转回 distance 格式
        "metadatas": [[{"page_num": chunk_map[r[0]]["page_num"],
                        "has_table": chunk_map[r[0]].get("has_table", False),
                        "chunk_index": chunk_map[r[0]]["chunk_index"]} for r in ranked]]
    }
    return results

# ---- 对比测试 ----
print("\n🔍 Dense vs Hybrid 对比:")
test_q = "What was Apple's net income in 2024?"

print(f"\nQ: {test_q}")
print(f"\n--- Dense only ---")
dense_res = search(test_q, collection, embed_model, top_k=5)
for i, (doc, dist) in enumerate(zip(dense_res["documents"][0], dense_res["distances"][0])):
    print(f"  [{i+1}] (sim={1-dist:.3f}) {doc[:100]}...")

print(f"\n--- Hybrid (BM25 + Dense) ---")
hybrid_res = hybrid_search(test_q, collection, embed_model, bm25, chunks, top_k=5)
for i, (doc, dist) in enumerate(zip(hybrid_res["documents"][0], hybrid_res["distances"][0])):
    print(f"  [{i+1}] (score={1-dist:.3f}) {doc[:100]}...")
```

```python
# Cell 3: Hybrid vs Dense 完整评估对比
from findoc_qa import rag_answer as dense_rag_answer
from openai import OpenAI

import os
OPENAI_API_KEY = os.environ.get("OPENAI_API_KEY", "")
openai_client = OpenAI(api_key=OPENAI_API_KEY)

def hybrid_rag_answer(question, collection, embed_model, bm25, chunks, top_k=5):
    """用 hybrid search 替代 dense search 的 RAG"""
    results = hybrid_search(question, collection, embed_model, bm25, chunks, top_k=top_k)
    docs = results["documents"][0]
    metas = results["metadatas"][0]
    distances = results["distances"][0]

    context_parts = []
    for i, (doc, meta, dist) in enumerate(zip(docs, metas, distances)):
        context_parts.append(f"[Source {i+1}, Page {meta['page_num']}] (relevance: {1-dist:.2f})\n{doc}")
    context = "\n\n---\n\n".join(context_parts)

    system_prompt = """You are a financial document analyst. Answer the user's question based ONLY on the provided context from the document.
Rules:
- Cite specific sources using [Source N] format
- If the context doesn't contain enough information, say so clearly
- For numerical data, quote exact figures from the document
- Be concise but thorough"""

    user_prompt = f"""Context from Apple 10-K filing:\n\n{context}\n\n---\n\nQuestion: {question}\n\nAnswer with citations:"""

    response = openai_client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "system", "content": system_prompt},
                  {"role": "user", "content": user_prompt}],
        temperature=0, max_tokens=800
    )
    answer = response.choices[0].message.content
    tokens_used = response.usage.total_tokens

    return {
        "question": question,
        "answer": answer,
        "sources": [{"page": m["page_num"], "distance": d, "text": doc[:200]}
                     for m, d, doc in zip(metas, distances, docs)],
        "tokens_used": tokens_used
    }

# Ground Truth
GROUND_TRUTH = [
    {"question": "What was Apple's total net sales in fiscal year 2024?", "keywords": ["391,035", "391"], "category": "numerical"},
    {"question": "How much did Apple spend on R&D in 2024?", "keywords": ["31,370", "31.4"], "category": "numerical"},
    {"question": "What was Apple's net income in 2024?", "keywords": ["93,736", "93.7"], "category": "numerical"},
    {"question": "What is Apple's total number of employees?", "keywords": ["164,000", "164"], "category": "numerical"},
    {"question": "Who is Apple's CEO?", "keywords": ["Tim Cook", "Cook"], "category": "factual"},
    {"question": "What was Apple's gross margin in 2024?", "keywords": ["46.2", "46%", "180,683"], "category": "numerical"},
    {"question": "How much cash did Apple return to shareholders in 2024?", "keywords": ["dividend", "repurchas", "buyback", "100 billion", "94,949", "15,234"], "category": "numerical"},
    {"question": "What are Apple's main product categories?", "keywords": ["iPhone", "Mac", "iPad", "Wearables", "Services"], "category": "factual"},
    {"question": "In which countries does Apple have retail stores?", "keywords": ["retail", "store", "country", "countries"], "category": "factual"},
    {"question": "What was iPhone revenue in 2024?", "keywords": ["201,183", "201"], "category": "numerical"},
    {"question": "What was Services revenue in 2024?", "keywords": ["96,169", "96"], "category": "numerical"},
    {"question": "What credit rating does Apple have?", "keywords": ["AA", "Aa1", "credit", "rating"], "category": "factual"},
]

def eval_keyword_hit(answer, keywords):
    return any(kw.lower() in answer.lower() for kw in keywords)

# 运行对比
print("📊 Dense vs Hybrid 完整评估\n")

dense_hits = 0
hybrid_hits = 0
results_comparison = []

for item in GROUND_TRUTH:
    q = item["question"]

    # Dense
    d_result = dense_rag_answer(q, collection, embed_model)
    d_hit = eval_keyword_hit(d_result["answer"], item["keywords"])
    dense_hits += int(d_hit)

    # Hybrid
    h_result = hybrid_rag_answer(q, collection, embed_model, bm25, chunks)
    h_hit = eval_keyword_hit(h_result["answer"], item["keywords"])
    hybrid_hits += int(h_hit)

    d_icon = "✅" if d_hit else "❌"
    h_icon = "✅" if h_hit else "❌"
    improved = " ⬆️" if (h_hit and not d_hit) else ""

    print(f"   Dense:{d_icon}  Hybrid:{h_icon}{improved}  {q}")

    results_comparison.append({
        "question": q,
        "dense_hit": d_hit,
        "hybrid_hit": h_hit,
        "dense_answer": d_result["answer"][:200],
        "hybrid_answer": h_result["answer"][:200]
    })

print(f"\n{'─' * 50}")
print(f"   Dense Accuracy:  {dense_hits}/{len(GROUND_TRUTH)} = {dense_hits/len(GROUND_TRUTH):.1%}")
print(f"   Hybrid Accuracy: {hybrid_hits}/{len(GROUND_TRUTH)} = {hybrid_hits/len(GROUND_TRUTH):.1%}")
improvement = hybrid_hits - dense_hits
if improvement > 0:
    print(f"   🎯 Hybrid improved {improvement} question(s)!")

with open("eval/hybrid_vs_dense.json", "w", encoding="utf-8") as f:
    json.dump(results_comparison, f, indent=2, ensure_ascii=False)
print(f"\n✅ Saved → eval/hybrid_vs_dense.json")
```

```python
# Cell 4: 调整 dense_weight 找最优融合比例
print("🔬 Dense weight 消融实验\n")

weights_to_test = [0.3, 0.4, 0.5, 0.6, 0.7, 0.8, 0.9]
weight_results = []

for w in weights_to_test:
    hits = 0
    for item in GROUND_TRUTH:
        result = hybrid_rag_answer(item["question"], collection, embed_model, bm25, chunks, top_k=5)
        # 需要重新用这个 weight 跑 hybrid_search
        pass  # 先用下面的方式

# 实际上需要把 weight 传进去，改一下：
for w in weights_to_test:
    hits = 0
    for item in GROUND_TRUTH:
        res = hybrid_search(item["question"], collection, embed_model, bm25, chunks,
                           top_k=5, dense_weight=w)
        docs = res["documents"][0]
        metas = res["metadatas"][0]
        distances = res["distances"][0]

        context_parts = []
        for i, (doc, meta, dist) in enumerate(zip(docs, metas, distances)):
            context_parts.append(f"[Source {i+1}, Page {meta['page_num']}]\n{doc}")
        context = "\n\n---\n\n".join(context_parts)

        response = openai_client.chat.completions.create(
            model="gpt-4o-mini",
            messages=[
                {"role": "system", "content": "You are a financial document analyst. Answer based ONLY on the provided context. Cite sources using [Source N]. For numerical data, quote exact figures. Be concise."},
                {"role": "user", "content": f"Context:\n\n{context}\n\n---\nQuestion: {item['question']}\nAnswer with citations:"}
            ],
            temperature=0, max_tokens=800
        )
        answer = response.choices[0].message.content
        hit = eval_keyword_hit(answer, item["keywords"])
        hits += int(hit)

    acc = hits / len(GROUND_TRUTH)
    print(f"   dense_weight={w:.1f} → Accuracy: {hits}/{len(GROUND_TRUTH)} = {acc:.1%}")
    weight_results.append({"dense_weight": w, "accuracy": acc, "hits": hits})

# 找最优
best = max(weight_results, key=lambda x: x["accuracy"])
print(f"\n🏆 Best: dense_weight={best['dense_weight']} → {best['accuracy']:.1%}")

with open("eval/weight_ablation.json", "w", encoding="utf-8") as f:
    json.dump(weight_results, f, indent=2)
print("✅ Saved → eval/weight_ablation.json")
```
